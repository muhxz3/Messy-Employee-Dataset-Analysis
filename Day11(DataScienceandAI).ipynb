#### MCAR - Missing completely at random
- The missingness has no relationship with the data.
#### MAR - Missing at random
- The missingness is related to other data or variables.
#### MNAR - Missing not at random
- The missingness is completely related to the missing value itself.

### Handling Missing Values

###### 1. Deletion
- Remove the missing observations/ entries
- using dropna() or dropnull() functions
- dropna(subset=["column_names"]) -- dropping rows
- dropna(axis=1) -- dropping rows
- <b>thresh</b> :- df.dropna(thresh=2) -- keep the rows that have atleast 2 non-null values
###### 2. Imputation
- Replace the missing observations or entries with a certain values (mean(), median(), mode(), etc.)
- df.fillna(value)<br>
eg:- df.fillna(0)<br>
df.fillna("N/A")
- fill different values for different columns:- <br>
df.fillna(
  {
    "marks":0,
    "city":"NA"
  }
)
- df["Marks"].fillna(df["marks"].mean())
- llly, median -- mid value
- mode -- the value that appears most of the times.<br>
df["city"].fillna(df["city"].mode())




---

###### forwardfill
- ffill() - fills missing entries with prev values
###### backwardfill
- bfill() - fills missing entries with next values



---





#### Duplicates
- df.duplicated()
- for quantifying:- <br>
df.duplicated().sum()

```python
# drop the missing values
data_drop = data.copy()
print("Original Data: ",data_drop.shape)

data_drop = data_drop.dropna()
print("After dropping Data: ",data_drop.shape)
```

```python
data_drop = data.dropna(subset=["City"])
print("After dropping Data: ",data_drop.shape)
```

```python
# thresh

data_thresh = data.dropna(thresh=7)
print(data_thresh.shape)
```

```python
# imputation part

data_fill = data.copy()
data_fill["Department"] = data_fill["Department"].fillna("Unknown")
data_fill["Department"].isna().sum()
```

```python
data_fill["Department"]
```

```python
data_fill.isna().sum()
```

```python
# fillna different values for different columns

data_fill = data.fillna({
    "Age": 18,
    "Annual_Salary": 0
})
```

```python
data_fill["Annual_Salary"].isna().sum()
```

```python
# imputation by mean, median or mode

df_mean = data.copy()
df_mean["Age"] = df_mean["Age"].fillna(df_mean["Age"].mean())
df_mean["Age"].isna().sum()
```
