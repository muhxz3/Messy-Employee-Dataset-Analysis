# Identifying Missing Values

```python
import pandas as pd
```

```python
# Loading the dataset
data = pd.read_csv("/content/Day11_Messy_Company_Employee_Dataset.csv")
data
```

```python
# identifying missing values
data.isna()
```

```python
data.isna().sum()
```

```python
data.isna().sum().sum()
```

- There are 32 Null/Missing Values

```python
data.dtypes
```

```python
data.shape
```

```python
# dropping rows with null values in Gender
data = data.dropna(subset=['Gender'])
data
```

```python
data.shape
```

```python
data.isna().sum()
```

```python
data
```

```python
# Filling Annual Salary
data['Annual_Salary'] = data['Annual_Salary'].fillna(data['Annual_Salary'].mean())
data
```

```python
data.isna().sum()
```

```python
# filling missing experience years with zero
data["Experience_Years"] = data["Experience_Years"].fillna(0)
```

```python
data.isna().sum()
```

```python
data.dtypes
```

```python
data["Age"]
```

```python
# dropping rows missing Age Values
data = data.dropna(subset=["Age"])
```

```python
data.shape
```

```python
data["Performance_Score"]
```

```python
# filling missing performance score with its mean
data["Performance_Score"] = data["Performance_Score"].fillna(data["Performance_Score"].mean())
```

```python
data.isna().sum()
```

```python
data["Work_Mode"]
```

```python
data["Work_Mode"].values
```

```python
# dropping rows with missing work modes
data = data.dropna(subset=["Work_Mode"])
```

```python
data.shape
```

```python
# filling missinf city to unknown
data["City"] = data["City"].fillna("Unknown")
```

```python
# dropping missing department values
data = data.dropna(subset=["Department"])
```

```python
data.isna().sum()
```

- all the missing values have been corrected


```python
data
```

# Identifying Inconsistent values

```python
data.groupby("Department")["Department"].value_counts()
```

```python
data.loc[data["Department"].isin(["ENGINEERING", "engineering"]), "Department"] = "Engineering"
```

```python
data.groupby("Department")["Department"].value_counts()
```

```python
data.groupby("Job_Title")["Job_Title"].value_counts()
```

```python
data.groupby("Gender")["Gender"].value_counts()
```

```python
data.loc[data["Gender"]=='female',"Gender"] = 'Female'
data.loc[data["Gender"]=='MALE',"Gender"] = 'Male'
```

```python
data.groupby("Gender")["Gender"].value_counts()
```

```python
data.groupby("City")["City"].value_counts()
```

```python
data.loc[data["City"].isin(["DELHI","delhi","Delhi "]),"City"] = "Delhi"
```

```python
data.groupby("City")["City"].value_counts()
```

```python
data.groupby("Work_Mode")["Work_Mode"].value_counts()
```

```python
data.loc[data["Work_Mode"].isin(["REMOTE","remote"]),"Work_Mode"] = 'Remote'
```

```python
data.groupby("Work_Mode")["Work_Mode"].value_counts()
```

```python
data.duplicated().sum()
```

```python
data = data.drop_duplicates()
data
```
